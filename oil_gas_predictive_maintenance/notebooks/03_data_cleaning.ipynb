# 03 — Data Cleaning

Applies the reusable cleaning pipeline in `src/data_preprocessing.py` to the raw sensor data: dtype fixes, invalid-value flagging, duplicate removal, group-wise imputation, and outlier winsorization.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..') / 'src'))

import pandas as pd
from data_preprocessing import clean_raw_dataframe, load_and_clean

raw = pd.read_csv('../data/raw/oil_gas_equipment_sensor_data.csv')
print('Raw shape:', raw.shape)
print('Raw missing values:', raw.isna().sum().sum())
print('Raw duplicate rows:', raw.duplicated().sum())
```

## Run the cleaning pipeline

```python
cleaned = load_and_clean(
    raw_path='../data/raw/oil_gas_equipment_sensor_data.csv',
    processed_path='../data/processed/oil_gas_equipment_cleaned.csv',
)
print('Cleaned shape:', cleaned.shape)
print('Remaining missing values:', cleaned.isna().sum().sum())
```

## Before / after comparison

```python
import plotly.express as px
import plotly.io as pio
pio.templates.default = 'plotly_dark'

comparison = pd.DataFrame({
    'Stage': ['Raw', 'Cleaned'],
    'Rows': [len(raw), len(cleaned)],
    'Missing_Values': [raw.isna().sum().sum(), cleaned.isna().sum().sum()],
})
comparison
```
