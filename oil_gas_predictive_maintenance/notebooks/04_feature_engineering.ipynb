# 04 — Feature Engineering

Builds industry-oriented engineered features on top of the cleaned dataset using `src/feature_engineering.py`.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..') / 'src'))

import pandas as pd
from feature_engineering import add_engineered_features, ENGINEERED_FEATURE_COLUMNS

df = pd.read_csv('../data/processed/oil_gas_equipment_cleaned.csv')
df_fe = add_engineered_features(df)
df_fe.to_csv('../data/processed/oil_gas_equipment_features.csv', index=False)
print(df_fe.shape)
df_fe[ENGINEERED_FEATURE_COLUMNS].head()
```

## Inspect the new engineered features

```python
df_fe[ENGINEERED_FEATURE_COLUMNS].describe().T
```

## Health Index vs actual failure outcome

```python
import plotly.express as px
import plotly.io as pio
pio.templates.default = 'plotly_dark'

px.box(df_fe, x='Failure_Within_30_Days', y='Health_Index', color='Failure_Within_30_Days',
       title='Engineered Health Index vs Actual Failure Outcome')
```

## Sensor anomaly count vs failure

```python
anomaly_fail = df_fe.groupby('Sensor_Anomaly_Count')['Failure_Within_30_Days'].mean().reset_index()
px.bar(anomaly_fail, x='Sensor_Anomaly_Count', y='Failure_Within_30_Days', title='Failure Rate by Sensor Anomaly Count')
```
